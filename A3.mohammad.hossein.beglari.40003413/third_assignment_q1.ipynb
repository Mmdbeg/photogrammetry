```python
import cv2 
import numpy
import matplotlib.pyplot as plt 
import numpy as np 
```

loading of first three images (1-2-3)

```python
img1 = cv2.imread("1.png", cv2.IMREAD_GRAYSCALE)
img2 = cv2.imread("2.png", cv2.IMREAD_GRAYSCALE)
img3 = cv2.imread("3.png", cv2.IMREAD_GRAYSCALE)
plt.subplot(131), plt.imshow(img1, cmap='gray')
plt.subplot(132), plt.imshow(img2, cmap='gray')
plt.subplot(133), plt.imshow(img3, cmap='gray')
plt.show()
```

Compute the 2D Fourier transform for each image and shifting  them to the center . //// and computing the magnitude of them 

```python
f1 = np.fft.fft2(img1)
f2 = np.fft.fft2(img2)
f3 = np.fft.fft2(img3)
fshift1 = np.fft.fftshift(f1)
fshift2 = np.fft.fftshift(f2)
fshift3 = np.fft.fftshift(f3)
magnitude_spectrum1 = np.log(np.abs(fshift1) + 1)
magnitude_spectrum2 = np.log(np.abs(fshift2) + 1)
magnitude_spectrum3 = np.log(np.abs(fshift3) + 1)
```

displaying the fourier transformation of images 

```python
plt.subplot(234), plt.imshow(magnitude_spectrum1, cmap='gray'), plt.title('Fourier Spectrum 1')
plt.subplot(235), plt.imshow(magnitude_spectrum2, cmap='gray'), plt.title('Fourier Spectrum 2')
plt.subplot(236), plt.imshow(magnitude_spectrum3, cmap='gray'), plt.title('Fourier Spectrum 3')
plt.show()
```
